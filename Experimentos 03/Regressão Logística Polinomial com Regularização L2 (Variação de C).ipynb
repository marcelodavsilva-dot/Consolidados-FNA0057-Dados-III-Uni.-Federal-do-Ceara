```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix 
from sklearn.inspection import DecisionBoundaryDisplay
```

```python
df = np.loadtxt("artificial2d.csv", delimiter = ",", skiprows = 1)
```

```python
np.random.seed(2)
np.random.shuffle(df)
```

```python
X = df[:,[0, 1]]
X = np.c_[X, X[:,0]**2, X[:,1]**2]
y = df[:,2]

n = X.shape[0]
m = y.shape[0]
```

```python
def standardize (V):

    df_padronizado = np.zeros(V.shape)

    for i in range (V.shape[1]):

      media = np.mean(V[:,i])
      desv = np.std(V[:,i])
      df_padronizado[:,i] = (V[:,i] - media) / desv

    return (df_padronizado)
```

```python
df_padronizado = standardize(X)
```

```python
X_treino =  df_padronizado[:int(0.7*n)]
X_teste = df_padronizado[int(0.7*n):]
y_treino = y[:int(0.7*m)]
y_teste = y[int(0.7*m):]
```

```python
regressao = LogisticRegression(penalty = None)

regressao.fit(X_treino, y_treino)
predicao = regressao.predict(X_teste)

print (predicao)
```

```python
cm = confusion_matrix(y_teste, predicao)

print("Matriz de Confusão:", f'\n')

vp = cm[1, 1]
print("Verdadeiro Positivo:", vp)

fn = cm[1, 0]
print("Falso Negativo:", fn)

fp = cm[0, 1]
print("Falso Positivo:", fp)

vn = cm[0, 0]
print("Verdadeiro Negativo:", vn, f'\n')

Accuracy = (vp + vn) / (vp + vn + fp + fn)
print ("Accuracy:", Accuracy, f'\n')

Recall = vp / (vp + fn)
print ("Recall:", Recall, f'\n')

Precision = vp / (vp + fp)
print ("Precision:", Precision, f'\n')
```

```python
C = [0.001, 0.01, 0.1, 1.0, 10.0, 100]

for c in C:
    regressao_c = LogisticRegression(penalty='l2', C=c)
    regressao_c.fit(X_treino, y_treino)
    predicao_c = regressao_c.predict(X_teste)

    cm_c = confusion_matrix(y_teste, predicao_c)

    vp = cm_c[1, 1]
    print("Métrica com C:", c, f'\n',f'\n',"Verdadeiro Positivo:", vp)

    fn = cm_c[1, 0]
    print("Falso Negativo:", fn)

    fp = cm_c[0, 1]
    print("Falso Positivo:", fp)

    vn = cm_c[0, 0]
    print("Verdadeiro Negativo:", vn, f'\n')

    Accuracy = (vp + vn) / (vp + vn + fp + fn)
    print ("Accuracy:", Accuracy, f'\n')

    Recall = vp / (vp + fn)
    print ("Recall:", Recall, f'\n')

    Precision = vp / (vp + fp)
    print ("Precision:", Precision, f'\n')
```
